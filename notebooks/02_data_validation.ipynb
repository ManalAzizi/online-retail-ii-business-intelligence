```python
import pandas as pd
from pathlib import Path

ventes = pd.read_parquet(Path("../data/processed") / "online_retail_II_cleaned.parquet")

print("Lignes :", len(ventes))
print("Lignes qui répètent (Invoice, StockCode) :", ventes.duplicated(["Invoice", "StockCode"]).sum())

par_facture = ventes.groupby("Invoice")[["Customer ID", "Country"]].nunique()
print("Factures avec plus d'un client :", (par_facture["Customer ID"] > 1).sum())
print("Factures avec plus d'un pays :", (par_facture["Country"] > 1).sum())

pays_par_client = ventes.groupby("Customer ID")["Country"].nunique()
print("Clients par nombre de pays :")
display(pays_par_client.value_counts().sort_index())

desc_par_produit = ventes.groupby("StockCode")["Description"].nunique()
print("Produits (StockCode) distincts :", ventes["StockCode"].nunique())
print("StockCode par nombre de descriptions :")
display(desc_par_produit.value_counts().sort_index())

print("Pays distincts :", ventes["Country"].nunique())
print("Jours distincts :", ventes["InvoiceDate"].dt.normalize().nunique())
print("Quantity min :", ventes["Quantity"].min(), "| Price min :", ventes["Price"].min())
print("Valeurs manquantes :", int(ventes[["Invoice", "StockCode", "Quantity", "Price",
      "Customer ID", "Country", "Montant"]].isna().sum().sum()))
```
